```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/ai4i2020.csv")

print(df.shape)
print(df.head())
print(df.info())
```

```python
df = df.rename(columns={
    "UDI": "udi",
    "Product ID": "product_id",
    "Type": "type",
    "Air temperature [K]": "air_temperature",
    "Process temperature [K]": "process_temperature",
    "Rotational speed [rpm]": "rotational_speed",
    "Torque [Nm]": "torque",
    "Tool wear [min]": "tool_wear",
    "Machine failure": "machine_failure"
})
```

```python
failure_mode_columns = [
    "TWF",
    "HDF",
    "PWF",
    "OSF",
    "RNF"
]
```

```python
features = [
    "type",
    "air_temperature",
    "process_temperature",
    "rotational_speed",
    "torque",
    "tool_wear"
]

X = df[features]
y = df["machine_failure"]
```

```python
print(df["machine_failure"].value_counts())
```

```python
failure_rate = df["machine_failure"].mean() * 100
print(f"Failure rate: {failure_rate:.2f}%")
```

```python
df["temperature_difference"] = (
    df["process_temperature"] -
    df["air_temperature"]
)
```

```python
df["power"] = (
    df["torque"] *
    df["rotational_speed"] *
    (2 * np.pi / 60)
)
```

```python
df["strain_indicator"] = (
    df["tool_wear"] *
    df["torque"]
)
```
